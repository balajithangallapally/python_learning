```python
user_name = "a"
password = 9889
while True:
    u_input = input("please enter the username :")
    p_input = int(input("please enter the password :"))
    if u_input == user_name and p_input == password:
        print("Valid input")
        break
    elif u_input == user_name and p_input != password:
        print("please enter valid password")
    else:
        print("plese enter valid username")      
```

```python
u = ["test","test1","test2","test3"]
p =["123","456","789","1011"]
while True:
    u_input = input("please enter the username :")
    p_input = input("please enter the password :")
    if u_input in u and p_input in p and u.index(u_input) == p.index(p_input):
        print("Success")
        break
    elif u_input in u and p_input not in p :
        print("please enter valid password")
    else:
        print("plese enter valid username")      
```
